```python
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels=['Young_Adults','Adults','Middle_Age','Senior']
df['group_age']=pd.qcut(df['age'], q=4,labels=labels)
df[['age','group_age']]
```

df['frequency_of_purchases']

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Annually': 365,
    'Bi-Weekly': 14,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Monthly': 30
}

# 1. Map the dictionary values to the new column
df['Frequency_day_purchase'] = df['frequency_of_purchases'].map(frequency_mapping)

# 2. FIX: View the correct columns side-by-side
df[['frequency_of_purchases', 'Frequency_day_purchase']]
```

```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Define connection parameters
USER = "root"
PASSWORD = "1234"
HOST = "localhost"  # Or "127.0.0.1"
PORT = "3306"  # Default MySQL port
DATABASE = "customer_behaviour"

# 2. Create the connection engine
connection_string = f"mysql+mysqlconnector://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
engine = create_engine(connection_string)

table_name='customer'
df.to_sql(table_name,engine,if_exists="replace",index=False)
print(f'data successfully loaded into table {table_name}, database {DATABASE}')

```

```python
# 3. Test the connection by running a query
query = "SELECT * FROM customer LIMIT 5;"
df = pd.read_sql(query, con=engine)

# 4. View the dataframe
df.head()
```
